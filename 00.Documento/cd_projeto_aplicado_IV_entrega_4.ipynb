# **Referencial Teórico**

O estudo do desemprego tem sido objeto central da economia do trabalho e das ciências sociais, dada sua relação direta com bem-estar social, crescimento econômico e segurança pública.

Segundo o IBGE (2024), o desemprego é definido como a situação de indivíduos em idade ativa que, apesar de estarem disponíveis e aptos para o trabalho, não encontram ocupação no mercado formal ou informal. Esse indicador é amplamente utilizado para avaliar a saúde da economia de um país.

Do ponto de vista social, a literatura aponta que a ausência de emprego contribui para a exclusão social, a degradação do capital humano e o aumento da criminalidade. Kahn (2020), em estudo para o Fórum Brasileiro de Segurança Pública, destaca que as taxas de violência estão fortemente correlacionadas com o desemprego, especialmente em contextos de vulnerabilidade social. De maneira semelhante, Santiago e Guimarães (2017) analisaram séries temporais e identificaram padrões consistentes de aumento da criminalidade em períodos de maior desocupação.

Na dimensão econômica, Lopes, Cavalcante e Oliveira (2020) reforçam que o desemprego provoca redução do consumo, queda da arrecadação tributária e comprometimento da capacidade de investimento público e privado. Esses efeitos, quando persistentes, ampliam desigualdades e travam o desenvolvimento sustentável.

A interseção entre desigualdade racial, gênero e escolaridade também é amplamente documentada. Pesquisas do IBGE mostram que mulheres e pessoas negras enfrentam taxas de desemprego superiores à média nacional, além de apresentarem menores rendimentos mesmo quando inseridos no mercado de trabalho. Essa dimensão interseccional reforça a necessidade de análises que vão além dos números agregados, evidenciando vulnerabilidades específicas que precisam ser enfrentadas por políticas públicas focalizadas.

Em síntese, o referencial teórico demonstra que o desemprego não é apenas um indicador econômico, mas um fenômeno multidimensional, cujos impactos sociais e econômicos se reforçam mutuamente. Esse entendimento fundamenta a presente pesquisa, que busca, por meio da análise dos dados da PNAD Contínua, oferecer subsídios técnicos e sociais para a formulação de políticas públicas mais eficazes.

Embora haja uma recuperação gradual a partir de 2022 e os níveis retornaram próximos aos patamares pré-2014, ainda temos alguns pontos de fragilidade, que merecem atenção e ação, pois mostram sensibilidade prolongada a choques macroeconômicos.

    9.2 Desigualdades de Gênero
A comparação entre homens e mulheres evidenciou uma diferença estrutural:
•	as mulheres apresentaram taxas maiores de desocupação, cerca de 30% superiores às masculinas em praticamente todos os anos observados.
Mesmo em fases de retomada econômica, essa assimetria persiste, influenciada por fatores como:
•	maior informalidade feminina,
•	interrupções de carreira associadas ao cuidado familiar,
•	dificuldades de reinserção no mercado de trabalho.
O comportamento consistente desse indicador reforça a necessidade de políticas específicas para mitigação dessas desigualdades.

    9.3 Nível de Instrução e Mercado de Trabalho
A análise por nível educacional evidenciou a relação inversa entre escolaridade e desocupação. A adaptação da classificação permitiu observar quatro grupos principais:
1.	Ensino Fundamental Incompleto ou sem instrução — maiores taxas de desocupação e maior instabilidade.
2.	Ensino Fundamental Completo / Ensino Médio Incompleto — redução moderada da vulnerabilidade, porém ainda elevados índices de informalidade.
3.	Ensino Médio Completo / Ensino Superior Incompleto — melhora na probabilidade de emprego, mas com forte exposição a ciclos econômicos.
4.	Ensino Superior Completo — menores taxas de desocupação e maiores rendimentos médios.
Além disso, observou-se que:
•	crises econômicas impactam de forma mais intensa os grupos com menor instrução;
•	trabalhadores qualificados sofrem com subutilização e desalinhamento entre formação e demanda.
A inclusão da variável renda reforça essa relação: escolaridade mais alta corresponde a maior rendimento médio, embora diferenças significativas persistam entre regiões e grupos demográficos.

    9.5 Disparidades Regionais
As regiões apresentaram comportamentos distintos e consistentes:
•	Nordeste e Norte mantiveram as maiores taxas de desocupação e os menores rendimentos médios;
•	Sul e Centro-Oeste registraram os melhores indicadores;
•	Sudeste apresentou forte oscilação, influenciada pela concentração de setores industriais sensíveis a crises.
Essas diferenças acompanham padrões históricos de desigualdade territorial, infraestrutura produtiva e grau de formalização do trabalho.

    9.6 Relação entre Desemprego e Violência
A análise integrada entre indicadores de violência (homicídios e latrocínios) e taxa de desocupação mostrou padrões relevantes:
•	em 2015–2016 e 2020, os picos de desemprego coincidiram com aumentos expressivos em crimes violentos;
•	unidades federativas com maiores oscilações econômicas apresentaram maior variabilidade nos índices de violência;
•	a correlação observada é consistente, embora não implique causalidade direta.
Os resultados sugerem que vulnerabilidade econômica e exclusão social ampliam riscos associados à criminalidade.

    9.7 Efeitos da Pandemia e Transformações Estruturais
O período pós-pandemia trouxe mudanças significativas:
•	aumento do trabalho remoto/híbrido em determinados setores;
•	desigual impacto entre grupos de baixa escolaridade, que dependeram majoritariamente de atividades presenciais;
•	ampliação das desigualdades regionais e educacionais;
•	queda temporária nas matrículas e interrupção de trajetórias escolares, especialmente no Ensino Médio.
Essas transformações ajudam a explicar parte das variações observadas nas taxas de desocupação e nos indicadores de renda e violência.

    9.8 Síntese Geral dos Resultados
Em conjunto, os resultados apontam que:
•	o desemprego no Brasil é multifatorial e altamente sensível a crises;
•	desigualdades de gênero e escolaridade têm papel central na explicação das diferenças observadas;
•	disparidades regionais permanecem determinantes na estrutura do mercado de trabalho;
•	a relação entre vulnerabilidade econômica e violência é consistente e deve ser considerada em políticas públicas.
Esses achados estruturam a base analítica necessária para formulação das recomendações apresentadas no capítulo seguinte.


# **Dicussão e Conclusão**

Com base nos resultados obtidos e nas relações identificadas entre desocupação, escolaridade, desigualdades sociais e indicadores de violência, são apresentadas recomendações de políticas públicas organizadas em três horizontes de atuação. O objetivo é oferecer intervenções coordenadas que reduzam o desemprego, promovam inclusão produtiva e atenuem vulnerabilidades estruturais no Brasil.

    10.1 Curto Prazo — Resposta Emergencial
Medidas destinadas a reduzir impactos imediatos de crises econômicas, choques sanitários ou aumento repentino da desocupação.

    10.1.1 Ampliação de programas de transferência de renda
Garantir subsistência mínima para famílias vulneráveis, com prioridade para:
•	pessoas com Ensino Fundamental incompleto;
•	jovens sem inserção no mercado;
•	regiões com altas taxas de desocupação.
    
    10.1.2 Incentivos fiscais temporários para micro e pequenas empresas
Reduzir custos operacionais e preservar empregos em setores mais sensíveis a períodos de retração econômica, além de incentivos fiscais, buscando pulverizar o mercado de trabalho, não o limitando a grandes centros, como as capitais e regiões metropolitanas.

    10.1.3 Programas de qualificação rápida
Oferecer capacitações de curta duração para trabalhadores impactados por crises, automação ou mudanças tecnológicas. A execução pode ser ampliada por meio do Sistema S (SENAI, SENAC, SESI, SENAR).

    10.1.4 Fortalecimento da intermediação de mão de obra
Modernizar e ampliar plataformas como o SINE, permitindo:
•	maior visibilidade de vagas;
•	orientação profissional;
•	integração com serviços digitais de recolocação.

    10.2 Médio Prazo — Requalificação e Estruturação
Medidas voltadas à reconstrução da capacidade produtiva e ao alinhamento da força de trabalho às demandas do mercado.

    10.2.1 Centros regionais de capacitação profissional
Criar centros integrando ensino técnico, universidades, Sistema S e setor privado para formação alinhada às necessidades locais.
Priorizar regiões com maiores taxas de desocupação e menor acesso à educação formal.
Incentivar regiões pouco desenvolvidas, par que o desenvolvimento econômico e social seja possível.

    10.2.2 Incentivo a setores de alta empregabilidade
Fomentar setores em expansão, como:
•	tecnologia da informação;
•	energia renovável;
•	economia verde;
•	infraestrutura digital.

O foco é gerar novas vagas e reduzir a dependência de setores mais vulneráveis a crises.

    10.2.3 Políticas de inclusão feminina no mercado de trabalho
Promover ações que reduzam desigualdades de gênero, como:
programas de retorno ao trabalho para mães;
•	ampliação de creches públicas;
•	incentivo à equidade salarial;
•	capacitações direcionadas a mulheres em situação de vulnerabilidade.

    10.2.4 Fomento à formalização e à economia solidária
Estimular modelos de trabalho cooperativo, microempreendedorismo e transição da informalidade para a formalidade, sobretudo nas regiões Norte e Nordeste.

    10.2.5 Expansão estruturada do ensino a distância (EAD)
Ampliar acesso à formação técnica e superior por meio do EAD, priorizando:
•	trabalhadores em áreas remotas;
•	pessoas com baixa escolaridade que precisam conciliar estudo e trabalho;
•	jovens que abandonaram os estudos durante crises econômicas.

    10.3 Longo Prazo — Transformação Estrutural
Ações contínuas voltadas à redução definitiva das desigualdades e ao fortalecimento da base produtiva nacional.

    10.3.1 Investimento permanente em educação básica e superior
Fortalecer a qualidade da educação com foco em:
•	alfabetização plena;
•	habilidades digitais;
•	pensamento crítico;
•	inovação e tecnologia.

O objetivo é elevar o nível de escolaridade e, como consequência, a renda média e a redução da violência.

    10.3.2 Redução de desigualdades raciais e de gênero
Reforçar políticas afirmativas, fiscalização trabalhista e programas que promovam inclusão e progressão profissional para grupos historicamente excluídos.

    10.3.3 Desenvolvimento regional equilibrado
Promover políticas de descentralização produtiva, como:
•	estímulo à interiorização de indústrias;
•	criação de polos tecnológicos em cidades médias;
•	infraestrutura integrada (transporte, conectividade, energia).
    
    10.3.4 Integração entre políticas de emprego e segurança pública
Articular estratégias que considerem o emprego como vetor de inclusão e redução da criminalidade, reforçando a relação entre:
•	mercado de trabalho mais forte;
•	menor vulnerabilidade social;
•	queda de violência (homicídios e latrocínios).

Considerações Finais – Versão Revisada e Aprimorada

O estudo analisou o desemprego no Brasil entre 2012 e 2025 sob uma perspectiva multidimensional, combinando mercado de trabalho, escolaridade, renda, diferenças regionais e indicadores de violência. Esses segmentos enfrentam as maiores taxas de desocupação e permanecem mais expostos à informalidade, à subutilização e à menor renda média, mesmo em períodos de recuperação econômica. A análise temporal evidenciou como os choques econômicos de 2015–2016 e a pandemia de 2020 provocaram rupturas significativas nas curvas de desocupação, com impactos mais fortes nos grupos de menor instrução e menor renda. No pós-pandemia, tendências como trabalho remoto, ensino híbrido e digitalização passaram a influenciar a dinâmica do emprego, ampliando oportunidades para alguns perfis, mas aprofundando desigualdades para quem não tem acesso a conectividade e qualificação digital.

Isso reforça a necessidade de integrar políticas de emprego, educação e segurança pública.
